```python
import pandas as pd
import os
from tqdm import tqdm
import warnings
import random
from nltk.tokenize.punkt import PunktSentenceTokenizer
from joblib import Parallel, delayed
import random

random.seed(123)
warnings.filterwarnings("ignore")
```

# US

```python
csv_path = "/hpc/dctrl/as1676/data/comp_ideology_detection/interventions_us.csv"
out_path = "/hpc/dctrl/as1676/data/comp_ideology_detection/training_set_us.csv"
```

```python
all_texts = []

df = pd.read_csv(csv_path)
all_texts.extend(df["speech"].astype(str).tolist())
```

```python
len(all_texts)
```

```python
def tokenize_doc(doc):
    return tokenizer.tokenize(doc)

train_sents = []
tokenizer = PunktSentenceTokenizer()
train_set = random.sample(all_texts, 200000)

results = Parallel(n_jobs=16)(
    delayed(tokenize_doc)(doc) for doc in train_set
)

train_sents = [sent for sub in results for sent in sub]
```

```python
out = pd.DataFrame({"sentence": train_sents})
out.to_csv(out_path, index=False)

print("Done! Saved", len(train_sents), "sentences.")
```
